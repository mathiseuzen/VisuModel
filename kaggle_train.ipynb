# Entraînement du CNN d'imagettes (VisuModel)

Avant de lancer :
1. À droite : **Add Input** → votre dataset d'imagettes (celui créé depuis `patches.zip`). Il doit apparaître sous `/kaggle/input`.
2. **Settings → Accelerator → GPU T4** (ou P100).
3. **Settings → Internet → On** (pour cloner le code depuis GitHub).
4. **Run All**. Le modèle apparaît dans l'onglet **Output** (`model_patches.pt`).

```python
!rm -rf VisuModel && git clone -q https://github.com/mathiseuzen/VisuModel.git
import torch
print('GPU :', torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'AUCUN (activez-le dans Settings)')
```

```python
import glob
import os
import zipfile

# 1. Ce qui a été ajouté en entrée (pour comprendre si ça ne marche pas)
for d, subs, files in os.walk('/kaggle/input'):
    if d.count('/') <= 4:
        print(d, '| dossiers :', subs[:8], '| fichiers :', files[:3])

# 2. Si Kaggle a laissé l'archive telle quelle, on la décompresse
for z in glob.glob('/kaggle/input/**/*.zip', recursive=True):
    print('décompression de', z)
    zipfile.ZipFile(z).extractall('/kaggle/working/data')

# 3. Dossier qui contient train/, valid/ et test/
data = next((d for root in ('/kaggle/input', '/kaggle/working/data')
             for d, subs, _ in os.walk(root) if {'train', 'valid', 'test'} <= set(subs)), None)
assert data, "Aucun dossier avec train/, valid/ et test/ : ajoutez votre dataset d'imagettes (Add Input)."
print('données :', data)
```

```python
!python VisuModel/train_patch_cnn.py --data {data} --epochs 40 --workers 2 --out /kaggle/working/model_patches.pt
```
